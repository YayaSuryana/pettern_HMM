```python
from hmmlearn import hmm
import numpy as np
import math

# Inisialisasi model HMM
model = hmm.MultinomialHMM(n_components=2)
model.startprob_ = np.array([0.6, 0.4])
model.transmat_ = np.array([[0.7, 0.3],
                            [0.4, 0.6]])
model.emissionprob_ = np.array([[0.1, 0.4, 0.5],
                                [0.6, 0.3, 0.1]])

# peluang observasi: 0 = walk, 1 = shop, 2 = clean

print("--- Hasil: Observasi 'walk' [0] ---")
print(math.exp(model.score(np.array([[0]]))))

print("\n--- Hasil: Observasi 'shop' [1] ---")
print(math.exp(model.score(np.array([[1]]))))

print("\n--- Hasil: Observasi 'clean' [2] ---")
print(math.exp(model.score(np.array([[2]]))))

print("\n--- Hasil: Observasi {'clean', 'clean'} [2, 2] ---")
print(math.exp(model.score(np.array([[2, 2]]))))



# viterbi algoritma

print("\n--- Hasil: Decode Viterbi [1, 2, 0] ---")
logprob1, seq1 = model.decode(np.array([[1, 2, 0]]).transpose())
print(math.exp(logprob1))
print(seq1)

print("\n--- Hasil: Decode Viterbi [2, 2, 2] ---")
logprob2, seq2 = model.decode(np.array([[2, 2, 2]]).transpose())
print(math.exp(logprob2))
print(seq2)

# Tugas Praktikum

print("\n--- Hasil: Latihan 1 (Peluang {'clean', 'clean', 'walk', 'walk', 'shop'}) ---")
print(math.exp(model.score(np.array([[2, 2, 0, 0, 1]]))))

print("\n--- Hasil: Latihan 2 (State optimal {'clean', 'clean', 'walk', 'walk', 'shop'}) ---")
logprob3, seq3 = model.decode(np.array([[2, 2, 0, 0, 1]]).transpose())
print(math.exp(logprob3))
print(seq3)
```
